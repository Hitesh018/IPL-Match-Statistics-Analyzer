```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.io as pio

pio.renderers.default = "browser"

from IPython.display import display

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# Plot settings
sns.set_theme(style="whitegrid")

# Load dataset
df = pd.read_csv("IPL_Cleaned.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("========== DATASET OVERVIEW ==========")

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
display(df.head())
```

```python
print("========== DATA TYPES ==========")

display(
    pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.astype(str),
        "Missing Values": df.isnull().sum().values,
        "Unique Values": df.nunique().values
    })
)
```

```python
missing = df.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

if len(missing) > 0:
    plt.figure(figsize=(12, 6))
    sns.barplot(x=missing.values, y=missing.index)
    plt.title("Missing Values by Column")
    plt.xlabel("Number of Missing Values")
    plt.ylabel("Column")
    plt.show()
else:
    print("No missing values found!")
```

```python
duplicates = df.duplicated().sum()

print("Number of Duplicate Records:", duplicates)

if duplicates == 0:
    print("No duplicate records found.")
else:
    print("Duplicate records need to be investigated.")
```

```python
print("========== STATISTICAL SUMMARY ==========")

display(df.describe().T)
```

```python
print("========== IPL SUMMARY ==========")

print("Total Matches:", df["match_id"].nunique())
print("Total Seasons:", df["season"].nunique())

teams = pd.unique(
    df[["batting_team", "bowling_team"]].values.ravel()
)

print("Total Teams:", len(teams))

print("Total Batters:", df["batter"].nunique())

if "bowler" in df.columns:
    print("Total Bowlers:", df["bowler"].nunique())

print("Total Runs:", df["runs_total"].sum())

if "bowler_wicket" in df.columns:
    print("Total Wickets:", int(df["bowler_wicket"].sum()))
```

```python
season_matches = (
    df.groupby("season")["match_id"]
    .nunique()
    .reset_index(name="matches")
)

plt.figure(figsize=(12, 5))

sns.barplot(
    data=season_matches,
    x="season",
    y="matches"
)

plt.title("Number of IPL Matches per Season")
plt.xlabel("Season")
plt.ylabel("Matches")
plt.xticks(rotation=45)

plt.show()
```

```python
# ==========================================
# Total Runs Per Season
# ==========================================

season_runs = (
    df.groupby("season", sort=False)["runs_total"]
    .sum()
    .reset_index()
)

# Convert season to string
season_runs["season"] = season_runs["season"].astype(str)

plt.figure(figsize=(12, 5))

plt.plot(
    season_runs["season"],
    season_runs["runs_total"],
    marker="o",
    linewidth=2
)

plt.title("Total Runs Scored per IPL Season")
plt.xlabel("Season")
plt.ylabel("Total Runs")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
numeric_cols = [
    "runs_batter",
    "runs_extras",
    "runs_total",
    "balls_faced",
    "bowler_wicket",
    "valid_ball"
]

available_cols = [
    col for col in numeric_cols
    if col in df.columns
]

correlation = df[available_cols].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of IPL Statistics")

plt.show()
```

```python
top_run_scorers = (
    df.groupby("batter")["runs_batter"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_run_scorers.values,
    y=top_run_scorers.index
)

plt.title("Top 10 IPL Run Scorers")
plt.xlabel("Runs")
plt.ylabel("Player")

plt.show()
```

```python
top_wicket_takers = (
    df.groupby("bowler")["bowler_wicket"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_wicket_takers.values,
    y=top_wicket_takers.index
)

plt.title("Top 10 IPL Wicket Takers")
plt.xlabel("Wickets")
plt.ylabel("Player")

plt.show()
```

```python
# ============================================
# TEAM PERFORMANCE HEATMAP
# ============================================

team_stats = (
    df.groupby("batting_team")
    .agg(
        Total_Runs=("runs_total", "sum"),
        Avg_Runs=("runs_total", "mean"),
        Matches=("match_id", "nunique")
    )
    .sort_values("Total_Runs", ascending=False)
)

plt.figure(figsize=(10, 7))

sns.heatmap(
    team_stats,
    annot=True,
    fmt=".0f",
    cmap="YlOrRd",
    linewidths=0.5
)

plt.title("Team Performance Heatmap")
plt.xlabel("Performance Metrics")
plt.ylabel("Teams")
plt.tight_layout()
plt.show()
```

```python
team_wickets = (
    df.groupby("bowling_team")["bowler_wicket"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=team_wickets.values,
    y=team_wickets.index
)

plt.title("Total Wickets by Team")
plt.xlabel("Wickets")
plt.ylabel("Team")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    df["runs_batter"],
    bins=10,
    kde=True
)

plt.title("Distribution of Runs Scored per Delivery")
plt.xlabel("Runs")
plt.ylabel("Frequency")

plt.show()
```

```python
# ==========================================
# Scoring Pattern by Over
# ==========================================

over_runs = (
    df.groupby("over")["runs_total"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 5))

plt.fill_between(
    over_runs["over"],
    over_runs["runs_total"],
    alpha=0.4
)

plt.plot(
    over_runs["over"],
    over_runs["runs_total"],
    marker="o"
)

plt.title("Average Scoring Pattern Across Overs")
plt.xlabel("Over")
plt.ylabel("Average Runs")
plt.tight_layout()

plt.show()
```

```python
over_runs = (
    df.groupby("over")["runs_total"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=over_runs,
    x="over",
    y="runs_total",
    marker="o"
)

plt.title("Average Runs by Over")
plt.xlabel("Over")
plt.ylabel("Average Runs")

plt.show()
```

```python
fig = px.bar(
    top_run_scorers.reset_index(),
    x="runs_batter",
    y="batter",
    orientation="h",
    title="Top 10 IPL Run Scorers"
)

fig.update_layout(
    xaxis_title="Runs",
    yaxis_title="Player"
)

fig.show()
```

```python
# ==========================================
# Player Performance
# ==========================================

player_stats = (
    df.groupby("batter")
    .agg(
        runs=("runs_batter", "sum"),
        balls=("balls_faced", "sum")
    )
)

player_stats["strike_rate"] = (
    player_stats["runs"] / player_stats["balls"] * 100
)

player_stats = player_stats[
    (player_stats["balls"] >= 100)
].copy()

plt.figure(figsize=(10, 6))

plt.scatter(
    player_stats["balls"],
    player_stats["runs"],
    s=50,
    alpha=0.6
)

plt.title("Player Runs vs Balls Faced")
plt.xlabel("Balls Faced")
plt.ylabel("Total Runs")

plt.tight_layout()
plt.show()
```
